## Mini Exploratory Data Analysis (EDA) on Titanic Dataset

**Objective:** Perform a deeper EDA on the Titanic dataset with better imputation, groupby-based insights, and multiple visualizations to tell a story.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic dataset directly from URL
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

df.head()
```

```python
print("Shape:", df.shape)
print("\nColumn Info:")
df.info()

print("\nMissing Values:")
df.isnull().sum()
```

## 1. Data Cleaning

```python
# Fill missing Age values with mean
df["Age"].fillna(df["Age"].mean(), inplace=True)

# Fill missing Embarked with mode (most frequent port)
df["Embarked"].fillna(df["Embarked"].mode()[0], inplace=True)

# Drop irrelevant / mostly-missing columns
df.drop(columns=["Cabin"], inplace=True)

# Confirm cleaning worked
print("Missing values after cleaning:")
df.isnull().sum()
```

## 2. Feature Engineering

```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 12, 18, 30, 50, 80],
    labels=["Child", "Teen", "YoungAdult", "Adult", "Senior"]
)

df[["Age", "AgeGroup"]].head()
```

```python
df["FamilySize"] = df["SibSp"] + df["Parch"]

df[["SibSp", "Parch", "FamilySize"]].head()
```

## 3. Analysis Questions

```python
age_group_survival = df.groupby("AgeGroup")["Survived"].mean().sort_values(ascending=False)
print("Survival Rate by Age Group:")
print(age_group_survival)
```

```python
embark_survival = df.groupby("Embarked")["Survived"].mean().sort_values(ascending=False)
print("Survival Rate by Embarkation Port:")
print(embark_survival)
```

```python
family_survival = df.groupby("FamilySize")["Survived"].mean().sort_values(ascending=False)
print("Survival Rate by Family Size:")
print(family_survival)
```

## 4. Visualizations

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Age"], bins=30, kde=True, color="steelblue")
plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8,6))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x="FamilySize", y="Survived", data=df, palette="viridis")
plt.title("Survival Rate by Family Size")
plt.xlabel("Family Size (SibSp + Parch)")
plt.ylabel("Survival Rate")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x="AgeGroup", y="Survived", data=df, palette="mako")
plt.title("Survival Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Survival Rate")
plt.show()
```

## Key Insights from EDA

- **Age Group:** Children had the highest survival rate, reflecting the "women and children first" evacuation priority.
- **Embarkation Port:** Passengers boarding from Cherbourg (C) had a higher survival rate compared to Southampton (S), possibly linked to class distribution at each port.
- **Family Size:** Passengers with small families (1-3 members) survived at higher rates than solo travelers or those with very large families, likely due to coordination difficulty in big families.
- **Correlation Heatmap:** `Pclass` showed a negative correlation with survival (lower class number = higher survival), while `Fare` showed a positive correlation.
